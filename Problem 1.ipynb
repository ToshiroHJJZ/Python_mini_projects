# Problem 1: Create a Funtion that allows one to recreate/crack a password given its length, without importing any python library and only using characters of the ASCII from A to Z and a to z

```python
def generate_hex_characters():
    hex_values = {}

    # Uppercase letters (A-Z)
    for char in range(ord('A'), ord('Z') + 1):
        hex_values[chr(char)] = hex(char)

    # Lowercase letters (a-z)
    for char in range(ord('a'), ord('z') + 1):
        hex_values[chr(char)] = hex(char)

    return hex_values

# Call the function and display the results
hex_characters = generate_hex_characters()

# Print the hexadecimal values
for char, hex_value in hex_characters.items():
    print(f"Character: {char}, Hexadecimal: {hex_value}")
```

```python
def get_hex_chars():
    chars = []
    for i in range(ord('A'), ord('Z') + 1):
        chars.append(chr(i))
    for i in range(ord('a'), ord('z') + 1):
        chars.append(chr(i))
    return chars

# Recursive function to generate all combinations
def pass_code(chars, length, prefix="", results=None):
    if results is None:
        results = []
    if length == 0:
        results.append(prefix)
    else:
        for c in chars:
            pass_code(chars, length - 1, prefix + c, results)
    return results

# Usage example:
length = 2  # Change to desired length
hex_chars = get_hex_chars()
combinations = pass_code(hex_chars, length)

# Print first 10 combinations (for large results)
for combo in combinations:
    print(combo)
```

How this works:

get_hex_chars() produces a list of all A-Z and a-z.

generate_combinations() recursively builds strings of desired length.
